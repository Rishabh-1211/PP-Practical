## Creating a Tuple

```python
# Creating a tuple

numbers = (1, 2, 3, 4, 5)

print(numbers)
print(type(numbers))
```

## Single Element Tuple

```python
# Single element tuple

single = (10,)

print(single)
print(type(single))
```

## Tuple Without Parentheses

```python
# Parentheses are optional when creating a tuple

numbers = 1, 2, 3, 4, 5

print(numbers)
print(type(numbers))
```

## Accessing Elements

```python
# Accessing elements

numbers = (1, 2, 3, 4, 5)

print("First element:", numbers[0])
print("Last element:", numbers[-1])
```

## Slicing

```python
# Slicing a tuple

numbers = (1, 2, 3, 4, 5)

print(numbers[1:4])
```

## Updating Elements

```python
# Tuples are immutable

numbers = (1, 2, 3, 4, 5)

# This will produce a TypeError
numbers[0] = 100
```

## Adding Elements

```python
# Adding an element to a tuple
# A new tuple is created

numbers = (1, 2, 3, 4, 5)

numbers = numbers + (6,)

print(numbers)
```

## Removing Elements

```python
# Removing an element from a tuple
# A new tuple is created

numbers = (1, 2, 3, 4, 5)

# Removing the element at index 2
numbers = numbers[:2] + numbers[3:]

print(numbers)
```

## Searching

### in operator

```python
# Searching for an element

numbers = (1, 2, 3, 2, 5)

print(3 in numbers)
print(10 in numbers)
```

### index()

```python
# Finding the position of an element

numbers = (1, 2, 3, 2, 5)

print(numbers.index(3))
```

### count()

```python
# Counting occurrences of an element

numbers = (1, 2, 3, 2, 5)

print(numbers.count(2))
```

## Sorting

### sorted()

```python
# Sorting a tuple

numbers = (4, 2, 5, 1, 3)

sorted_numbers = tuple(sorted(numbers))

# sorted() returns a list first, so we use tuple() to convert it back to a tuple

print("Original tuple:", numbers)
print("Sorted tuple:", sorted_numbers)
```

### sorted(reverse=True)

```python
numbers = (4, 2, 5, 1, 3)

reverse_sorted = tuple(sorted(numbers, reverse=True))

print("Original tuple:", numbers)
print("Reverse Sorted tuple:", reverse_sorted)
```

## Reversing a Tuple

```python
# Reversing a tuple

numbers = (1, 2, 3, 4, 5)

reversed_tuple = numbers[::-1]

print("Original tuple:", numbers)
print("Reversed tuple:", reversed_tuple)
```

## Copying a Tuple

```python
# Copying a tuple

numbers = (1, 2, 3, 4, 5)

copy_tuple = numbers

print("Original:", numbers)
print("Copy:", copy_tuple)
```

### Copy Using tuple()

```python
# Copying using tuple()

numbers = (1, 2, 3, 4, 5)

copy_tuple = tuple(numbers)

print("Original:", numbers)
print("Copy:", copy_tuple)
```

## Concatenation

```python
# Concatenating two tuples

a = (1, 2)
b = (3, 4)

result = a + b

print(result)
```

## Repetition

```python
# Repeating a tuple

a = (1, 2)

result = a * 3

print(result)
```

## Traversing a Tuple

```python
fruits = ("apple","banana","orange")
for x in fruits:
    print(x)
```
