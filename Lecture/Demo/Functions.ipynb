## Built-In Functions

```python
# Built-in functions are already available in Python

print("Hello Python")   # print() displays output

length = len("Python")  # len() returns the number of characters
print(length)

maximum = max(5, 10, 15)  # max() finds the largest value
print(maximum)
```

## User-Defined Function

```python
# Creating a function to find the square of a number

def square(n):
    return n * n

# Calling the function
print(square(5))
#square(6)
#square(7)
```

## Lambda Function

```python
# Lambda function is a small anonymous function
# It takes one value and returns its square

square = lambda x,y : x * y

# Calling the lambda function
print(square(4,7))
```

## Creating Functions in Python

## Function Without Parameters

```python
# Function without any parameters

def greet():
    # Printing a welcome message
    print("Hello, Welcome to Python!")

# Calling the function
greet()
```

## Function With Parameter

```python
# Function with a parameter

def greet(name):
    # name receives the value passed during function call
    print("Hello,", name)

# Passing "Alice" as an argument
greet("Alice")
```

## Function With Return Value

```python
# Function to add two numbers

def add(a, b):
    # Returning the sum of two numbers
    return a + b

# Calling the function and storing the result
result = add(10, 20)

# Printing the returned value
print(result)
```

## Types of Function Arguments

## Positional Arguments

```python
# Values are passed according to their position

def info(name, age):
    print("Name:", name)
    print("Age:", age)

# "John" goes to name and 25 goes to age
info("John", 25)
```

## Keyword Arguments

```python
# Values are passed using parameter names

def info(name, age):
    print("Name:", name)
    print("Age:", age)

# Order does not matter when using keyword arguments
info(age=25, name="John")
```

## Default Arguments

```python
# Function with a default value

def greet(name="Guest"):
    # If no value is given, "Guest" is used
    print("Hello", name)

# Using the default value
greet()

# Providing a new value
greet("Emma")
```

## Variable-Length Arguments

```python
# *numbers allows multiple values to be passed

def total(*numbers):
    # sum() adds all values received
    return sum(numbers)

# Passing multiple arguments
print(total(1, 2, 3, 4))
```

## Recursive Function

### Factorial Using Recursion

```python
# A recursive function calls itself

def factorial(n):
    # Base condition: stops recursion
    if n == 0 or n == 1:
        return 1

    # Function calls itself with a smaller value
    return n * factorial(n - 1)

# Calling the factorial function
print(factorial(5))
```
