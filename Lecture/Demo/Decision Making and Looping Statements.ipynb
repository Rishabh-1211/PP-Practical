## if statement

```python
age = 15

if age >= 18:
    print("Eligible to vote")
```

## if-else statement

```python
marks = 35

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

## nested if-else statement

```python
balance = 15000
amount = 11000

if balance >= amount:
    if amount <= 10000:
        print("Transaction Successful")
    else:
        print("Withdrawal Limit Exceeded")
else:
    print("Insufficient Balance")
```

## if-elif-else statement

```python
marks = 55

if marks > 100 or marks < 0:   
    print("Invalid Input")
elif marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

## switch-case (match-case)

```python
day = 3

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid")
```

## for loop

A for loop in Python is used to iterate over a sequence (such as a list, tuple, string, or range)

### for loop with range

```python
for i in range(10):
    print(i)
```

### for loop with list

```python
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
for fruit in fruits:
    print(fruit)
```

### for loop with string

```python
name = "Sridham"

for i in name:
    print(i)
```

## while loop

```python
count = 1

while count <= 5:
    print("Ritesh")
    count += 1
```

## break statement

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

## continue statement

```python
for i in range(10):
    if i == 3:
        continue
    print(i)
```

## pass statement

```python
for i in range(5):
    if i == 3:
        pass
    print(i)
```
