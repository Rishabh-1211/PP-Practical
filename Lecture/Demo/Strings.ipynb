# String 

### Creating Strings  

```python
name = "Python" 
text = 'Programming' 

print(name)
print(text)
```

```python
message = """This is 
a multi-line 
string."""


print(message)
```

### Accessing Characters 


```python
name = "Python" 
print(name[5]) 
print(name[-1])    
```

### Slicing 

```python
name = "Python Programming" 

print(name[1:11:2])    # yth 

print(name[:3])     

print(name[3:])     

print(name[::-1])   
```

### Updating Strings 
Strings are immutable set of characters. 

```python
name = "Python" 

# TypeError 
name[0] = "J" 
```

To modify a string, create a new one. 

```python
name = "J" + name[1:] 
print(name)
```

### Concatenation 

```python
first = "Hello" 
second = "World" 

print(first + " every " +second) 
# Hello World
```

### Repetition 

```python
print("Hi" * 3) 
# Hi Hi Hi 
```

### Searching 

```python
text = "Python Programming" 

print("Py" in text) 

print(text.find("b")) 

print(text.index("t")) 

print(text.count("m")) 
```

### Length 

```python
text = "Python" 
print(len(text)) 
```

### Changing Case 

```python
text = "python Programming" 

print(text.upper()) 

print(text.lower()) 

print(text.title()) 

print(text.capitalize()) 

print(text.swapcase()) 
```

### Removing Spaces 

```python
text = "   Python   " 

print(text.strip()) 
print(text.lstrip()) 
print(text.rstrip()) 
```

### Replacing Text 

```python
text = "I like Java" 
print(text.replace("Java", "Python")) 
```

### Splitting Strings 

```python
text = "Apple,Banana,Mango" 
print(text.split(",")) 
# ['Apple', 'Banana', 'Mango'] 
```

### Joining Strings 

```python
fruits = ["Apple", "Banana", "Mango"] 
print(", ".join(fruits)) 
# Apple, Banana, Mango 
```

### Starts With / Ends With 

```python
text = "Python Programming" 
print(text.startswith("Python")) 
print(text.endswith("ing")) 
```

### Checking String Type 

```python
text = " " 

print(text.isalpha())      # Letters only 

print(text.isdigit())      # Digits only 

print(text.isalnum())      # Letters & digits 

print(text.islower()) 

print(text.isupper()) 

print(text.isspace())
```

# Formatting Strings 


### Using f-strings (Recommended) 


```python
name = "John" 
age = 20 

print(f"My name is {name} and I am {age} years old.") 
```

### Using format() 

```python
name = "John" 
age = 20 

print("My name is {} and I am {} years old.".format(name, age)) 
```

### Using % Operator (Older Method) 


```python
name = "John" 
age = 20 

print("My name is %s and I am %d years old." % (name, age)) 
```

### Escape Characters 

```python
print("Hello\nWorld") 

print("Hello\tWorld") 

print("She said \"Hello\"") 

print("C:\\Users\\John") 
```

### String Comparison 

```python
print("apple" == "apple") 

print("apple" == "Apple") 

print("apple" < "banana") 
```

### Copying a String

text = "Python" 

copy_text = text

```python
print(copy_text)
```
