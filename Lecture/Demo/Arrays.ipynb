#### Creating Numpy Arrays

```python
a = np.array([1,2,3])
print(a)

b = np.zeros(5)
print(b)

c = np.ones(4)
print(c)

d = np.arange(1,10)
print(d)

e = np.linspace(1,20,10) 
print(e)
```

#### Array Attributes

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
```

```python
print(arr.ndim)

print(arr.shape)

print(arr.size)

print(arr.dtype)
```

#### Mathematical Operations on Array using Numpy 

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)

print(a-b)

print(a*b)

print(a/b)
```

#### Statistical Operations on Array using Numpy 

```python
arr = np.array([10,20,30,40])

print(np.sum(arr))

print(np.mean(arr))

print(np.max(arr))

print(np.min(arr))

print(np.std(arr))
```

```python
arr = np.arange(12)

print(arr)

new = arr.reshape(3,4)

print(new)
```
