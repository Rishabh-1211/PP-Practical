## Creating a List

```python
numbers = [1, 2, 3, 4, 5]

print(numbers)
```

## Accessing Elements

```python
numbers = [1, 2, 3, 4, 5]

print(numbers[0])      # First element
print(numbers[-1])     # Last element
```

## Slicing

```python
numbers = [1, 2, 3, 4, 5]

print(numbers[1:4])    # Elements from index 1 to 3
print(numbers[:2])     # First two elements
print(numbers[3:])     # From index 3 to end
```

## Adding Elements 

### append()

```python
numbers = [1, 2, 3, 4, 5]

numbers.append(6)

print(numbers)
```

### insert()

```python
numbers = [1, 2, 3, 4, 5]

numbers.insert(2, 10)

print(numbers)
```

### extend()

```python
numbers = [1, 2, 3, 4, 5]

numbers.extend([6, 7, 8])

print(numbers)
```

## Removing Elements

### remove()

```python
numbers = [1, 2, 3, 4, 5]

numbers.remove(3)

print(numbers)
```

### pop()

```python
numbers = [1, 2, 3, 4, 5]

numbers.pop()

print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]

a = numbers.pop()

print(numbers)

print(a)
```

### pop(index)

```python
numbers = [1, 2, 3, 4, 5]

numbers.pop(1)

print(numbers)
```

### del

```python
numbers = [1, 2, 3, 4, 5]

del numbers[0]

print(numbers)
```

### clear

```python
numbers = [1, 2, 3, 4, 5]

print(numbers)

numbers.clear()

print(numbers)
```

## Updating Elements

```python
numbers = [1, 2, 3, 4, 5]

numbers[0] = 100

print(numbers)
```

## Searching

### in operator

```python
numbers = [1, 2, 3, 4, 5]

print(3 in numbers)
print(10 in numbers)
```

### index()

```python
numbers = [1, 2, 3, 4, 5]

print(numbers.index(3))
```

### count()

```python
numbers = [1, 2, 3, 2, 4, 2, 5]

print(numbers.count(2))
```

## Sorting

### sort()

```python
numbers = [5, 2, 8, 1, 4]

numbers.sort()

print(numbers)
```

### sort(reverse=True)

```python
numbers = [5, 2, 8, 1, 4]

numbers.sort(reverse=True)

print(numbers)
```

### reverse()

```python
numbers = [1, 2, 3, 4, 5]

numbers.reverse()

print(numbers)
```

## Copying List 

### copy()

```python
numbers = [1, 2, 3, 4, 5]

copy_list = numbers.copy()

print(numbers)
print(copy_list)
```

### Copy using Slicing

numbers = [1, 2, 3, 4, 5]

copy_list = numbers[:]

print(copy_list)

## len()

```python
numbers = [1, 2, 3, 4, 5]

print(len(numbers))
```

## Concatenation

```python
a = [1, 2]
b = [3, 4]

result = a + b

print(result)
```

## Repitition

```python
numbers = [1, 2]

result = numbers * 3

print(result)
```
